```python
import pandas as pd
import matplotlib.pyplot as plt

results_df = pd.read_csv("results/abc_simulation_500_normal.csv")
plt.hist(
    results_df["p_AFR"],
    weights=results_df["posterior"],
    bins=20
)
plt.xlabel("p_AFR")
plt.ylabel("Posterior weight")
plt.title("Posterior distribution of p_AFR")

plt.show()
```
