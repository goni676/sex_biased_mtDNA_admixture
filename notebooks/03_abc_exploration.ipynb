```python
import pandas as pd
from pathlib import Path

from src.parameters import PARAMETER_VECTORS
from src.abc_simulation import simulate_theta_replicates
```

```python
prior_type = "normal"
n_replicates = 100
n_vectors = 500
```

```python
results = []

for vector_id in range(0, n_vectors):
    tmp = simulate_theta_replicates(PARAMETER_VECTORS, n_replicates, prior_type=prior_type)
    results.append({
        "vector_id": vector_id,
        **tmp[0],
        "p_EUR": 1-tmp[0]['p_AFR'],
        "likelihood": tmp[1],
        })

results_df = pd.DataFrame(results)
results_df.columns = results_df.columns.str.replace('parameters.', '', regex=False)

total_likelihood = results_df['likelihood'].sum()
if total_likelihood > 0:
    results_df['posterior'] = results_df['likelihood'] / (total_likelihood)
else:
    results_df['posterior'] = 0.0
    
results_dir = Path(__file__).resolve().parent.parent / "results"
results_dir.mkdir(exist_ok=True)

output_file = results_dir / f"abc_simulation_{n_vectors}_{prior_type}.csv"
results_df.to_csv(output_file, index=False)

print(f"Saved to: {output_file}")
```
